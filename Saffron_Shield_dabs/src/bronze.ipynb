```python
dbutils.widgets.text("catalog", "claims", "Catalog Name")
catalog = dbutils.widgets.get("catalog")
```

```python
# Load plans.csv
spark.sql(f"""
CREATE TABLE IF NOT EXISTS IDENTIFIER('{catalog}.raw.plans')
""")

spark.sql(f"""
COPY INTO IDENTIFIER('{catalog}.raw.plans')
FROM (
    SELECT *
    FROM '/Volumes/claims/raw/landing/plans/'
)
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true', 'inferSchema' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true')
""")
```
